```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.special import erf
```

# Unsteady uni-directional flow
Suppose a plate on the plane $z=0$ is forced from stationary at time $t=0$ to moving with speed $c$ for $t>0$.

We found that the resulting velocity is given as
$$ u(z,t) = c\bigg(1-\mathrm{erf}\bigg(\frac{z}{2\sqrt{\nu t}}\bigg)\bigg)$$

```python
# define velocity function
def u(z,t,c,nu):
    u = c*(1-erf(z/(2*np.sqrt(nu*t))))
    return u
```

```python
# choose parameters
c = 2.
nu = 0.1/0.91 #approximate kinematic viscosity of olive oil

z = np.linspace(0,10,50)
t = np.linspace(0,50,50)

Z,T = np.meshgrid(z,t)

vel = u(Z,T,c,nu)
```

## Visualise the velocity vector

```python
W = vel*0. # for plotting using quiver function
X = Z*0. # for plotting using quiver function
```

```python
fig = plt.figure(figsize=(12,4))
ax1, ax2, ax3 = fig.subplots(1, 3)

ax1.quiver(X[9,:],Z[9,:],vel[9,:],W[9,:], scale = 1, scale_units='x')#scale=1/np.sqrt(nu*5), 
ax1.set_xlabel('$x$')
ax1.set_ylabel('$z$')
ax1.set_title('$t=10$')
ax1.axis([-0.01,3,-0.01,10])

ax2.quiver(X[29,:],Z[29,:],vel[29,:],W[29,:], scale = 1, scale_units='x') #scale=1/np.sqrt(nu*15))
ax2.set_xlabel('$x$')
ax2.set_ylabel('$z$')
ax2.set_title('$t=30$')
ax2.axis([-0.01,3,-0.01,10])

ax3.quiver(X[49,:],Z[49,:],vel[49,:],W[49,:], scale = 1, scale_units='x') #scale=1/np.sqrt(nu*25))
ax3.set_xlabel('$x$')
ax3.set_ylabel('$z$')
ax3.set_title('$t=50$')
ax3.axis([-0.01,3,-0.01,10])

plt.show()
```
