```python
import matplotlib.pyplot as plt
import numpy as np
```

# Conformal mapping

## Flow in a wedge
Let's first plot the boundaries of the wedge

```python
# choose angle
alpha = np.pi/9

# define boundary on x-axis
bound_low = np.linspace(0,1,100)

# define boundary at angle alpha
bound_up = bound_low*np.tan(alpha)
```

```python
plt.plot(bound_low,bound_low*0,'k',linewidth=3)
plt.plot(bound_low,bound_up,'k',linewidth=3)
plt.xlabel('$x$')
plt.ylabel('$y$')
plt.title('$z$-plane')
plt.show()
```

Consider the mapping introduced in class, 
$$\tau(z) = z^\frac{\pi}{\alpha} $$
First, we need to define the boundaries as complex coordinates.

```python
# complex lower boundary (note - Python uses j as imaginary number instead of i)
bound_low_complex = bound_low + 1j*0.*bound_low

# define boundary on y-axis
bound_up_complex = bound_low + 1j*bound_up
```

Ensure this worked correctly by plotting real and imaginary parts of complex boundaries

```python
plt.plot(np.real(bound_low_complex),np.imag(bound_low_complex),'k',linewidth=3)
plt.plot(np.real(bound_up_complex),np.imag(bound_up_complex),'k',linewidth=3)
plt.xlabel('$\\Re(z)$')
plt.ylabel('$\\Im(z)$')
plt.title('$z$-plane')
plt.show()
```

Now map the boundaries to the $\tau$-plane

```python
# map lower boundary
bound_low_mapped = (bound_low_complex)**(np.pi/alpha)

# map upper boundary
bound_up_mapped = (bound_up_complex)**(np.pi/alpha)
```

```python
plt.plot(np.real(bound_low_mapped),np.imag(bound_low_mapped),'k',linewidth=3)
plt.plot(np.real(bound_up_mapped),np.imag(bound_up_mapped),'k',linewidth=3)
plt.xlabel('$\\Re(\\tau)$')
plt.ylabel('$\\Im(\\tau)$')
plt.title('$\\tau$-plane')
plt.axis([-2, 1.2, -0.02, 0.4])
plt.show()
```

Plot boundaries in different colors to visualise individual mapping

```python
fig = plt.figure(figsize=(12,4))
ax1, ax2 = fig.subplots(1, 2)

ax1.plot(np.real(bound_low_complex),np.imag(bound_low_complex),'purple',linewidth=3)
ax1.plot(np.real(bound_up_complex),np.imag(bound_up_complex),'turquoise',linewidth=3)
ax1.set_xlabel('$x$')
ax1.set_ylabel('$y$')
ax1.set_title('$z$-plane')
ax1.axis([-2, 1.2, -0.02, 0.4])

ax2.plot(np.real(bound_low_mapped),np.imag(bound_low_mapped),'purple',linewidth=3)
ax2.plot(np.real(bound_up_mapped),np.imag(bound_up_mapped),'turquoise',linewidth=3)
ax2.set_xlabel('$\\xi$')
ax2.set_ylabel('$\\eta$')
ax2.set_title('$\\tau$-plane')
ax2.axis([-2, 1.2, -0.02, 0.4])

plt.show()
```

For extra fun, track some specific points in the mapping.

```python
# define points in the z-plane
A = bound_low_complex[-1]
O = 0. + 1j*0.
B = bound_up_complex[-1]

# map to tau-plane
A_mapped = A**(np.pi/alpha)
O_mapped = O**(np.pi/alpha)
B_mapped = B**(np.pi/alpha)
```

```python
fig = plt.figure(figsize=(12,4))
ax1, ax2 = fig.subplots(1, 2)

ax1.plot(np.real(bound_low_complex),np.imag(bound_low_complex),'purple',linewidth=3)
ax1.plot(np.real(bound_up_complex),np.imag(bound_up_complex),'turquoise',linewidth=3)
ax1.set_xlabel('$x$')
ax1.set_ylabel('$y$')
ax1.set_title('$z$-plane')
ax1.axis([-2, 1.2, -0.02, 0.4])

ax1.plot(np.real(A),np.imag(A),'ks',ms=12)
ax1.plot(np.real(O),np.imag(O),'ko',ms=12)
ax1.plot(np.real(B),np.imag(B),'kd',ms=12)

ax2.plot(np.real(bound_low_mapped),np.imag(bound_low_mapped),'purple',linewidth=3)
ax2.plot(np.real(bound_up_mapped),np.imag(bound_up_mapped),'turquoise',linewidth=3)
ax2.set_xlabel('$\\xi$')
ax2.set_ylabel('$\\eta$')
ax2.set_title('$\\tau$-plane')
ax2.axis([-2, 1.2, -0.02, 0.4])

ax2.plot(np.real(A_mapped),np.imag(A_mapped),'ks',ms=12)
ax2.plot(np.real(O_mapped),np.imag(O_mapped),'ko',ms=12)
ax2.plot(np.real(B_mapped),np.imag(B_mapped),'kd',ms=12)

plt.show()
```

## Exploring other mappings

### Translation
Consider the map $\tau(z) = z + a $

```python
# Define z as upper half-plane
xs = np.linspace(-5,5,100)
ys = np.linspace(0,5,50)

X,Y = np.meshgrid(xs,ys)

Z = X+1j*Y

# define translation
a = 3.+1j*2.
tau_trans = Z + a
```

```python
fig = plt.figure(figsize=(12,4))
ax1, ax2 = fig.subplots(1, 2)

# plot original as contour
ax1.contourf(np.real(Z),np.imag(Z),np.abs(Z),20)
ax1.set_xlabel('$x$')
ax1.set_ylabel('$y$')
ax1.set_title('$z$-plane')

# plot map as contour
ax2.contourf(np.real(tau_trans),np.imag(tau_trans),np.abs(tau_trans),20)
ax2.set_xlabel('$\\xi$')
ax2.set_ylabel('$\\eta$')
ax2.set_title('$\\tau$-plane')

plt.show()
```

### Rotation
Consider the map $\tau(z) = z\mathrm{e}^{i\alpha} $

```python
# define rotation
tau_rot = Z*np.exp(1j*alpha)
```

```python
fig = plt.figure(figsize=(12,4))
ax1, ax2 = fig.subplots(1, 2)

# plot original as contour
ax1.contourf(np.real(Z),np.imag(Z),np.abs(Z),20)
ax1.set_xlabel('$x$')
ax1.set_ylabel('$y$')
ax1.set_title('$z$-plane')

# plot map as contour
ax2.contourf(np.real(tau_rot),np.imag(tau_rot),np.abs(tau_rot),20)
ax2.set_xlabel('$\\xi$')
ax2.set_ylabel('$\\eta$')
ax2.set_title('$\\tau$-plane')

plt.show()
```

### Reflection
Consider the map $\tau(z) = \bar{z} $

```python
# define reflection
tau_ref = np.real(Z)-1j*np.imag(Z)
```

```python
fig = plt.figure(figsize=(12,4))
ax1, ax2 = fig.subplots(1, 2)

# plot original as contour
ax1.contourf(np.real(Z),np.imag(Z),np.abs(Z),20)
ax1.set_xlabel('$x$')
ax1.set_ylabel('$y$')
ax1.set_title('$z$-plane')

# plot map as contour
ax2.contourf(np.real(tau_ref),np.imag(tau_ref),np.abs(tau_ref),20)
ax2.set_xlabel('$\\xi$')
ax2.set_ylabel('$\\eta$')
ax2.set_title('$\\tau$-plane')

plt.show()
```

### Infinite half-plane to finite circle
Consider the map ,
$$\tau(z) = \frac{z-i}{z+i} $$

```python
# define map
tau_circ = (Z-1j)/(Z+1j)
```

```python
fig = plt.figure(figsize=(12,4))
ax1, ax2 = fig.subplots(1, 2)

# plot original as contour
ax1.contourf(np.real(Z),np.imag(Z),np.abs(Z),20)
ax1.set_xlabel('$x$')
ax1.set_ylabel('$y$')
ax1.set_title('$z$-plane')

# plot map as contour
ax2.contourf(np.real(tau_circ),np.imag(tau_circ),np.abs(tau_circ),20)
ax2.set_xlabel('$\\xi$')
ax2.set_ylabel('$\\eta$')
ax2.set_title('$\\tau$-plane')

plt.show()
```

Looks like we might need to add more points in our upper half-plane

```python
xs_ext = np.linspace(-20,20,1000)
ys_ext = np.linspace(0,100,1000)

X_ext,Y_ext = np.meshgrid(xs_ext,ys_ext)

Z_ext = X_ext+1j*Y_ext

# define map
tau_circ_ext = (Z_ext-1j)/(Z_ext+1j)
```

```python
fig = plt.figure(figsize=(12,4))
ax1, ax2 = fig.subplots(1, 2)

# plot original as contour
ax1.contourf(np.real(Z_ext),np.imag(Z_ext),np.abs(Z_ext),20)
ax1.set_xlabel('$x$')
ax1.set_ylabel('$y$')
ax1.set_title('$z$-plane')

# plot map as contour
ax2.contourf(np.real(tau_circ_ext),np.imag(tau_circ_ext),np.abs(tau_circ_ext),20)
ax2.set_xlabel('$\\xi$')
ax2.set_ylabel('$\\eta$')
ax2.set_title('$\\tau$-plane')

plt.show()
```
